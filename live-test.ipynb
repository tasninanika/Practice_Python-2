# Write a Program to take a user score and find a student's grade using Python code. Following is the grade information.


```python
score = int(input("Enter the student score between (0-100): "))

if 90 <= score <= 100:
  print("Student score is A")
elif 80 <= score <= 89:
  print("Student score is B")
elif 70 <= score <= 79:
  print("Student score is C")
elif 60 <= score <= 69:
  print("Student score is D")
elif 50 <= score <= 59:
  print("Student score is E")
elif 40 <= score <= 49:
  print("Student score is E-")
elif 0 <= score < 40:
  print("Student score is F or Fail")
else:
  print("Wrong input.")
```
